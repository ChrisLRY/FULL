## 1.4 Data sets Splitting

**(1) Split data into training, validation, and test sets**

```python
# Perform stratified sampling to split data into training, validation, and test sets

# First split: training + validation (80%) and test (20%)
bigtrain_set, test_set = train_test_split(data, test_size=0.2, random_state=42, stratify=data['SC'])

# Second split: training (80% of 80% = 64%) and validation (20% of 80% = 16%)
train_set, val_set = train_test_split(bigtrain_set, test_size=0.2, random_state=42, stratify=bigtrain_set['SC'])
```

```python
# Calculate and display class distributions for each dataset
for dataset, name in zip([train_set, val_set, test_set], ['Train', 'Validation', 'Test']):
    distribution = dataset['SC'].value_counts(normalize=True).reset_index()
    distribution.columns = ['SC', 'Proportion']
    distribution['Weather'] = distribution['SC'].map(synop_mapping)
    distribution = distribution[['SC', 'Weather', 'Proportion']]  # Reorder columns
    print(f"\n{name} Set Distribution:")
    print(distribution.to_string(index=False))  # Print without the index
```

```python
# Define the feature columns
feature_columns = continuous_features + categorical_features  

# Extract features for training, validation, and test sets
X_train = train_set[feature_columns]
X_val = val_set[feature_columns]
X_test = test_set[feature_columns]

# Extract FSO_Att and RFL_Att as separate output sets for training, validation, and test
y_train_FSO = train_set['FSO_Att']
y_train_RF = train_set['RFL_Att']

y_val_FSO = val_set['FSO_Att']
y_val_RF = val_set['RFL_Att']

y_test_FSO = test_set['FSO_Att']
y_test_RF = test_set['RFL_Att']
```

```python
# Verify and summarize the shapes of the datasets
print(f"X_train: {X_train.shape}, y_train_FSO: {y_train_FSO.shape}, y_train_RF: {y_train_RF.shape}")
print(f"X_val: {X_val.shape}, y_val_FSO: {y_val_FSO.shape}, y_val_RF: {y_val_RF.shape}")
print(f"X_test: {X_test.shape}, y_test_FSO: {y_test_FSO.shape}, y_test_RF: {y_test_RF.shape}")
```

```python
# Create directories to save the files if they don't already exist
os.makedirs("Data/Splitting/Train/Generic/X", exist_ok=True)
os.makedirs("Data/Splitting/Train/Generic/Y", exist_ok=True)
os.makedirs("Data/Splitting/Validation/Generic/X", exist_ok=True)
os.makedirs("Data/Splitting/Validation/Generic/Y", exist_ok=True)
os.makedirs("Data/Splitting/Test/Generic/X", exist_ok=True)
os.makedirs("Data/Splitting/Test/Generic/Y", exist_ok=True)

# Save feature sets (X)
X_train.to_csv("Data/Splitting/Train/Generic/X/X_train.csv", index=False)
X_val.to_csv("Data/Splitting/Validation/Generic/X/X_val.csv", index=False)
X_test.to_csv("Data/Splitting/Test/Generic/X/X_test.csv", index=False)

# Save FSO_Att output sets (Y)
y_train_FSO.to_csv("Data/Splitting/Train/Generic/Y/y_train_FSO.csv", index=False)
y_val_FSO.to_csv("Data/Splitting/Validation/Generic/Y/y_val_FSO.csv", index=False)
y_test_FSO.to_csv("Data/Splitting/Test/Generic/Y/y_test_FSO.csv", index=False)

# Save RFL_Att output sets (Y)
y_train_RF.to_csv("Data/Splitting/Train/Generic/Y/y_train_RF.csv", index=False)
y_val_RF.to_csv("Data/Splitting/Validation/Generic/Y/y_val_RF.csv", index=False)
y_test_RF.to_csv("Data/Splitting/Test/Generic/Y/y_test_RF.csv", index=False)

print("Generic Datasets saved successfully.")
```

**(2) Creating SYNOP CODE-Based Subsets for Training and Validation Sets**

```python
# Dictionary to store subsets for each SYNOP CODE in training and validation sets
train_subsets = {code: subset for code, subset in train_set.groupby('SC')}
val_subsets = {code: subset for code, subset in val_set.groupby('SC')}

# Print the sizes of each subset for verification
print("Training subsets by SYNOP CODE:")
for code, subset in train_subsets.items():
    print(f"SYNOP CODE {code}: {len(subset)} samples")

print("\nValidation subsets by SYNOP CODE:")
for code, subset in val_subsets.items():
    print(f"SYNOP CODE {code}: {len(subset)} samples")
```

```python
# Dictionary to store feature and output subsets for each SYNOP CODE in training and validation sets
train_subsets_X = {}
train_subsets_y_FSO = {}
train_subsets_y_RF = {}

val_subsets_X = {}
val_subsets_y_FSO = {}
val_subsets_y_RF = {}

# Redefine feature_columns to ensure SC is not included
feature_columns_specific = [col for col in feature_columns if col != 'SC']

# Process each SYNOP CODE subset for the training set
for code, subset in train_subsets.items():
    # Extract features and outputs for the training subset
    train_subsets_X[code] = subset[feature_columns_specific] 
    train_subsets_y_FSO[code] = subset['FSO_Att']
    train_subsets_y_RF[code] = subset['RFL_Att']

# Process each SYNOP CODE subset for the validation set
for code, subset in val_subsets.items():
    # Extract features and outputs for the validation subset
    val_subsets_X[code] = subset[feature_columns_specific]
    val_subsets_y_FSO[code] = subset['FSO_Att']
    val_subsets_y_RF[code] = subset['RFL_Att']
```

```python
# Create directories to save the files if they don't already exist
os.makedirs("Data/Splitting/Train/Specific/X", exist_ok=True)
os.makedirs("Data/Splitting/Train/Specific/Y", exist_ok=True)
os.makedirs("Data/Splitting/Validation/Specific/X", exist_ok=True)
os.makedirs("Data/Splitting/Validation/Specific/Y", exist_ok=True)
os.makedirs("Data/Splitting/Test/Specific/X", exist_ok=True)
os.makedirs("Data/Splitting/Test/Specific/Y", exist_ok=True)

# Save training subsets for each SYNOP CODE
for code in train_subsets_X.keys():
    # Save features
    train_subsets_X[code].to_csv(f"Data/Splitting/Train/Specific/X/train_X_SC_{code}.csv", index=False)
    # Save FSO_Att output
    train_subsets_y_FSO[code].to_csv(f"Data/Splitting/Train/Specific/Y/train_y_FSO_SC_{code}.csv", index=False)
    # Save RFL_Att output
    train_subsets_y_RF[code].to_csv(f"Data/Splitting/Train/Specific/Y/train_y_RF_SC_{code}.csv", index=False)

# Save validation subsets for each SYNOP CODE
for code in val_subsets_X.keys():
    # Save features
    val_subsets_X[code].to_csv(f"Data/Splitting/Validation/Specific/X/val_X_SC_{code}.csv", index=False)
    # Save FSO_Att output
    val_subsets_y_FSO[code].to_csv(f"Data/Splitting/Validation/Specific/Y/val_y_FSO_SC_{code}.csv", index=False)
    # Save RFL_Att output
    val_subsets_y_RF[code].to_csv(f"Data/Splitting/Validation/Specific/Y/val_y_RF_SC_{code}.csv", index=False)


print("Specific Datasets saved successfully.")
```
